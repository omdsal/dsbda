```python
from scipy import stats
z=np.abs(stats.zscore(df['Emp_Marks']))
print(z)
```

```python
threshold = 0.22
sample_outliers=np.where(z<threshold)
sample_outliers
```

```python
new_df = df.copy()

# Dropping outliers from new_df
for i in sample_outliers:
    new_df.drop(i, inplace=True, errors='ignore')

# Display the updated new_df
new_df
```
